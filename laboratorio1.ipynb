```python
# Datos que vienen desprolijos del sistema de ventas
cliente_raw = "   mArIa gOnZaLeZ   "
codigo_producto = "prod-9948-laptop"
email_raw = "  MARIA.GONZALEZ@GMAIL.COM  "

# Limpieza y estandarización
cliente_clean = cliente_raw.strip().title()
producto_clean = codigo_producto.replace("prod-9948-", "").upper()
email_clean = email_raw.strip().lower()

print("Cliente:", cliente_clean)
print("Producto:", producto_clean)
print("Email:", email_clean)
```

```python
# Los precios a veces vienen como texto desde archivos CSV
precio_str = "1250.80"
cantidad_str = "5"
porcentaje_iva = 0.21

# Convertimos los tipos de datos (Typecasting)
precio = float(precio_str)
cantidad = int(cantidad_str)

# Cálculos de la operación
subtotal = precio * cantidad
monto_iva = subtotal * porcentaje_iva
total_final = subtotal + monto_iva

print("Subtotal acumulado:", subtotal)
print("Monto IVA (21%):", monto_iva)
print("Total bruto:", total_final)
```

```python
# Queremos saber cuántas cajas de 2 unidades necesitamos para empaquetar
unidades_por_caja = 2

# División entera para saber cajas llenas
cajas_completas = cantidad // unidades_por_caja

# Módulo para saber cuántas unidades quedan sueltas
unidades_sueltas = cantidad % unidades_por_caja

print("Cajas completas requeridas:", cajas_completas)
print("Unidades que van sueltas:", unidades_sueltas)
```

```python
# Redondeo de decimales y reporte final con f-strings
reporte = f"""
========================================
         REPORTE DE PROCESAMIENTO
========================================
Cliente: {cliente_clean}
Contacto: {email_clean}
Producto: {producto_clean}
----------------------------------------
Unidades: {cantidad} ({cajas_completas} cajas + {unidades_sueltas} suelta)
Total a Facturar: ${total_final:.2f}
========================================
"""

print(reporte)
```
